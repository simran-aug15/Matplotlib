```python
# import
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#  simple Bar chart
children = [10,20,40,10,30]
colors = ['red','blue','green','yellow','pink']
plt.bar(colors,children)
```

```python
# color
plt.bar(colors,children,color='black')
```

```python
# using Horizontal bar chart : Use this onluy then categories on x axis are more than Y axis
plt.barh(colors,children,color='black')
```

```python
# colors and labels : 
df=pd.read_csv('Sample.csv/batsman_season_record.csv')
df
plt.bar(df['batsman'],df['2015'])
```

```python
# Now if you want to print runs of 2015,2016,2017,using same code you can do it but they will overlap among each other ,
# but we donot want that ,so we use this next code instead for better understanding
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.2)
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2)  #df.shape[0] : It gives the rows in dataset, df.shape[1] : It gives the columns in dataset
plt.bar(np.arange(df.shape[0])+0.2,df['2017'],width=0.2)
plt.xticks(np.arange(df.shape[0]),df['batsman'])
plt.show()
```

```python
# a problem
children = [10,20,40,10,30]
colors =['red red red red red','blue blue blue blue blue','green green green green','yellow yellow yellow yellow','black black black black ']
plt.bar(colors,children,color='black')


# This code will give problem as values are large and give you no proper data, To make it work properly
plt.xticks(rotation='vertical')
plt.show()
```

### Stacked Bar chart

```python
plt.bar(df['batsman'],df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016')
plt.bar(df['batsman'],df['2017'],bottom=df['2015']+df['2016'],label='2017')
plt.legend()
plt.show()
```
