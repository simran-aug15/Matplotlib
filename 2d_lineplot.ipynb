```python
# from a pandas dataframe 
batsman=pd.read_csv('Sample.csv/sharma-kohli.csv')
batsman
plt.plot(batsman['index'],batsman['V Kohli'])


```

```python
# plotting multiple plot 
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
# label titleplt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.title('Rohit sharma VS Virat Kohli carrer Comparison')
plt.xlabel('Season')
plt.ylabel('Runs score')
```

```python
# Color(hex),line (width and style) and marker(size)
plt.plot(batsman['index'],batsman['V Kohli'],color='green')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black')
plt.title('Rohit sharma VS Virat Kohli carrer Comparison')
plt.xlabel('Season')
plt.ylabel('Runs score')

```

```python
# Using linestyle
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='dotted')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#F52746',linestyle='dashed')
plt.title('Rohit sharma VS Virat Kohli carrer Comparison')
plt.xlabel('Season')
plt.ylabel('Runs score')

```
