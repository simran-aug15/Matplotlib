### Matplotlib is a popular Python library used for creating charts, graphs, and data visualizations. It helps in representing data visually through plots like line charts, bar graphs, histograms, scatter plots, and more.


```python
# import 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### 2D line plot : Use for Bivariate analysis(apply on two columns)
### Categorical -> Numerical and Numerical ->Numerical 
### Use case : Time series data

```python
# plotting a simple function
price=[48000,54000,57000,49000,47000,45000]
year=[2015,2016,2017,2018,2019,2020]
plt.plot(year,price)
```

```python
# from a pandas dataframe 
batsman=pd.read_csv('Sample.csv/sharma-kohli.csv')
batsman
plt.plot(batsman['index'],batsman['V Kohli'])


```

```python
# plotting multiple plot 
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
# label titleplt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.title('Rohit sharma VS Virat Kohli carrer Comparison')
plt.xlabel('Season')
plt.ylabel('Runs score')
```

```python
# Color(hex),line (width and style) and marker(size)
plt.plot(batsman['index'],batsman['V Kohli'],color='green')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black')
plt.title('Rohit sharma VS Virat Kohli carrer Comparison')
plt.xlabel('Season')
plt.ylabel('Runs score')

```

```python
# Using linestyle
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='dotted')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#F52746',linestyle='dashed')
plt.title('Rohit sharma VS Virat Kohli carrer Comparison')
plt.xlabel('Season')
plt.ylabel('Runs score')

```

```python
# linewidth : To change the width of lines 
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='dotted',linewidth=3)
plt.plot(batsman['index'],batsman['RG Sharma'],color='#F52746',linestyle='dashed',linewidth=2)
plt.title('Rohit sharma VS Virat Kohli carrer Comparison')
plt.xlabel('Season')
plt.ylabel('Runs score')

```

```python
# marker : Make a difference marks in peak and points 
# marker : 'D','O','+','-'
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='dotted',marker='+',markersize=10)
plt.plot(batsman['index'],batsman['RG Sharma'],color='#F52746',linestyle='dashed',marker='+',markersize=8)
plt.title('Rohit sharma VS Virat Kohli carrer Comparison')
plt.xlabel('Season')
plt.ylabel('Runs score')

```

```python
# legend : location 
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='dotted',label="Virat")
plt.plot(batsman['index'],batsman['RG Sharma'],color='#F52746',linestyle='dashed',label="sharma")
plt.title('Rohit sharma VS Virat Kohli carrer Comparison')
plt.xlabel('Season')
plt.ylabel('Runs score')
plt.legend()

```

```python
# Before using limiting axes
price=[48000,54000,57000,49000,47000,45000,450000]
year=[2015,2016,2017,2018,2019,2020,2021]
plt.plot(year,price)
```

```python
#  After using limiting axes
price=[48000,54000,57000,49000,47000,45000,450000]
year=[2015,2016,2017,2018,2019,2020,2021]
plt.plot(year,price)
plt.ylim(0,100000)  # This is used when we only want to see the variation of less data,not the highest data
plt.xlim(2017,2019)
```

```python
# grid : To increase the readability
plt.plot(batsman)
plt.grid()
```

```python
# show : This remove all the extra information only show graph
plt.plot(batsman['index'],batsman['V Kohli'],color='green')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black')
plt.title('Rohit sharma VS Virat Kohli carrer Comparison')
plt.xlabel('Season')
plt.ylabel('Runs score')
plt.show()
```
