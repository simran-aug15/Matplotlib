```python
# import 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# import dataset
batter=pd.read_csv('Sample.csv/batter (1).csv')
batter
```

```python
# A different way to plot a graph
fig,ax=plt.subplots(figsize=(15,6))
ax.scatter(batter['avg'],batter['strike_rate'])
ax.set_title("something")
ax.set_xlabel('Avg')
ax.set_ylabel('strike_rate')
fig.show()

```

```python
fig,ax = plt.subplots(figsize=(15,6))

ax.scatter(batter['avg'],batter['strike_rate'],color='red',marker='+')
ax.set_title('Something')
ax.set_xlabel('Avg')
ax.set_ylabel('Strike Rate')

fig.show()
```

```python
# We are using index to plot multiple graph side by side
fig, ax = plt.subplots(nrows=2,ncols=1,sharex=True,figsize=(10,6))

ax[0].scatter(batter['avg'],batter['strike_rate'],color='red') #1st graph
ax[1].scatter(batter['avg'],batter['runs']) #2nd graph

ax[0].set_title('Avg Vs Strike Rate')
ax[0].set_ylabel('Strike Rate')


ax[1].set_title('Avg Vs Runs')
ax[1].set_ylabel('Runs')
ax[1].set_xlabel('Avg')
```
