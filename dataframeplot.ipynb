```python
# import
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

#### Dataframeplot : It is a function in pandas by which you can plot matplotlib graph in pandas

```python
# on a series
S=pd.Series([1,2,3,4,5,6,7])
S.plot(kind='line')
S.plot(kind='pie')
```

```python
# can be used on dataframe also
tips=sns.load_dataset('tips')
tips['size']=tips['size']*100
tips.head()
```

```python
# scatter plot -> labels -> markers ->figsize ->color ->cmap
tips.plot(kind='scatter',x='total_bill',y='tip',title='cost Analysis',marker='+',figsize=(10,6),s='size',c='sex',cmap='viridis')
```

```python
# 2d plot
stocks=pd.read_csv('https://raw.githubusercontent.com/m-mehdi/pandas_tutorials/main/weekly_stocks.csv')
stocks.head()
```

```python
# line plot on this(Series)
stocks['MSFT'].plot(kind='line')
```

```python
# Lineplot on dataframe
stocks.plot(kind='line',x='Date')
```

```python
stocks[['Date','MSFT','FB']].plot(kind='line')
```

```python
# bar charts ->single ->horizontal ->multiple
# using tips
temp=pd.read_csv('Sample.csv/batsman_season_record (1).csv')
temp.head()
temp['2015'].plot(kind='bar')
```

```python
tips.groupby('sex')['total_bill'].mean().plot(kind='bar')
```

```python
# side by side graph
temp.plot(kind='bar')
```

```python
# stacked bar charts
temp.plot(kind='bar',stacked='True')
```

```python
# Histogram
stocks.plot(kind='hist')
```

```python
stocks[['MSFT','FB']].plot(kind='hist',bins=40)
```

```python
# pie -> single and multiple
df=pd.DataFrame({'batsman':['Dhawan','Rohit','Kohli','SKY','Pandya','Pant'],
        'match1':[120,90,35,45,12,10],
        'match2':[0,1,123,130,34,45],
        'match3':[50,24,145,45,10,90]})
df.head()
```

```python
df['match1'].plot(kind='pie',labels=df['batsman'],autopct='%01f%%')
```

```python
# multiple pie charts
df[['match1','match2','match3']].plot(kind='pie',subplots=True,figsize=(15,8))
```

```python
# multiple separate graphs together
stocks.plot(kind='line',subplots=True)
```

```python
# On multiindex dataframe
tips.pivot_table(index='day',columns='sex',values='total_bill',aggfunc='mean').plot(kind='line')
```
