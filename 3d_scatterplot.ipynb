```python
# import 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#importing dataset
batter=pd.read_csv('Sample.csv/batter.csv')
batter
```

```python
#3d_scatter
fig=plt.figure()
ax=plt.subplot(projection='3d')
ax.scatter3D(batter['runs'],batter['avg'],batter['strike_rate'])
ax.set_title('IPL batsman analysis')
ax.set_xlabel('Runs')
ax.set_ylabel('Avg')
ax.set_zlabel('SR')
```

```python
# 3d line plot : You want to plot a line between the graph you plot
x=[0,1,5]
y=[0,10,13]
z=[0,13,20]
fig=plt.figure()
ax=plt.subplot(projection='3d')
ax.scatter3D(x,y,z,s=[100,100,100])  #s=increase the size of the point
ax.plot3D(x,y,z)  # join the points with line
```
