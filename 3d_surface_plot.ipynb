```python
# import 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 1st example:
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)
xx,yy=np.meshgrid(x,y)
yy.shape
z=xx**2 + yy**2
fig=plt.figure(figsize=(12,8))
ax=plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
# 2nd example :
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)
xx,yy=np.meshgrid(x,y)
z=np.sin(xx)+np.cos(yy)
fig=plt.figure(figsize=(12,8))
fig=plt.figure(figsize=(12,8))
ax=plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
# contour : 3D graph represents in 2D 
# code for the 1st example to show in 2d 
fig=plt.figure(figsize=(12,8))
ax=plt.subplot()
p=ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
# contour : 3D graph represents in 2D 
# code for the 2nd example to show in 2d 
fig=plt.figure(figsize=(12,8))
ax=plt.subplot()
p=ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
# Heatmap : A heatmap in Matplotlib is a graphical representation of data where values are shown using different colors.
#Higher values → darker/brighter colors
#Lower values → lighter/different colors
delivery=pd.read_csv('Sample.csv/IPL_Ball_by_Ball_2008_2022.csv')
delivery.head()
temp_df=delivery[(delivery['ballnumber'].isin([1,2,3,4,5,6])&(delivery['batsman_run']==6))]
grid=temp_df.pivot_table(index='overs',columns='ballnumber',values='batsman_run',aggfunc='count')
plt.figure(figsize=(20,10))
plt.imshow(grid)  #It shows the photos
plt.xticks(np.arange(0,6),list(range(1,7)))
plt.yticks(np.arange(0,20),delivery['overs'].unique())
plt.colorbar()
```
