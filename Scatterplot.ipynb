```python
# import 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# plt.scatter simple function
X=np.linspace(-10,10,50)
Y=10*X + 3 +np.random.randint(0,300,50)
Y
plt.scatter(X,Y)
```

```python
# plt.scatter on pandas data 
df=pd.read_csv('Sample.csv/batter.csv')
df.head(50)
plt.scatter(df['avg'],df['strike_rate'],marker='+')
plt.title('Avg an strike rate analysis of Top 50 Batsman')
plt.xlabel('Average')
plt.ylabel('SR')
```

```python
# size : Give the idea of how many people are there in data at that point 
# use people : small dot, more people : large
tips=sns.load_dataset('tips')
tips
plt.scatter(tips['total_bill'],tips['tip'],s=tips['size']*20)
```
